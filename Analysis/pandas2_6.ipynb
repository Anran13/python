```python
import pandas as pd
import numpy as np
```

```python
tips_df = pd.read_csv('tips.csv')
print(tips_df.head(3))
print(tips_df.info())

tips_df['smoker'].unique()

# print(tips_df[['smoker', 'day', 'time']].columns)

unique_values = {col: tips_df[col].unique() for col in tips_df[['smoker', 'day', 'time']].columns}
print(unique_values)

tips_df['proportion'] = tips_df['tip'] / tips_df['total_bill']
print(tips_df.info())
```

```python
# calculate count, mean, max for tip and total_bill
group_df = tips_df.groupby(by=['smoker', 'day'])[['tip', 'total_bill']].agg(['count', 'mean', 'max'])
print(group_df)
```
