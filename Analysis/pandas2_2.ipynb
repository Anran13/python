```python
tw2303 = yf.download('2303.TW', start=start_date, end=end_date,auto_adjust=True)
tw2454 = yf.download('2454.TW', start=start_date, end=end_date,auto_adjust=True)  
tw2317 = yf.download('2317.TW', start=start_date, end=end_date,auto_adjust=True)
```
