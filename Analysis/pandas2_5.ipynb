```python
df = pd.read_csv('各鄉鎮市區人口密度.csv', skiprows=1)
print(df.info)
print(df.columns)
```

```python
df.columns = ['year', 'district', 'population', 'area', 'density']
print(df.tail(3))
print(df.shape)
```

```python
df1 = df.dropna(axis=0, how='any')
print(df1.head(3))
print(df1.tail(3))
print(df1.shape)
```

```python
df2 = df1.drop('year', axis=1)
print(df2.head(5))
print(df2.info())
```

```python
print(df2.nunique())
unique_values = df2.apply(lambda x: x.unique())
print(unique_values)
```

```python
def find_duplicate(df:pd.DataFrame, col_names:str):
    counts = df.groupby(col_names).size()
    dup_values = counts[counts > 1].index
    duplicates = df[df[col_names].isin(dup_values)]
    print(duplicates)


find_duplicate(df2, 'population')

col_names = [col for i, col in enumerate(df.columns) if i != 2]
print(col_names)

print(col_names[2])


# dup_flags = df2[col_names].apply(lambda col: col.duplicated(keep=False))
# print(dup_flags)

```
