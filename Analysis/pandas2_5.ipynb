```python
df = pd.read_csv('各鄉鎮市區人口密度.csv', skiprows=1)
print(df.info)
print(df.columns)
```

```python
df.columns = ['year', 'district', 'population', 'area', 'density']
print(df.tail(3))
print(df.shape)
```

```python
df1 = df.dropna(axis=0, how='any')
print(df1.head(3))
print(df1.tail(3))
print(df1.shape)
```

```python
df2 = df1.drop('year', axis=1)
print(df2.head(5))
print(df2.info())
```

```python
print(df2.nunique())
unique_values = df2.apply(lambda x: x.unique())
print(unique_values)
```

```python
def find_duplicate(df:pd.DataFrame, col_names:str):
    counts = df.groupby(col_names).size()
    dup_values = counts[counts > 1].index
    duplicates = df[df[col_names].isin(dup_values)]
    print(duplicates)

# find_duplicate(df2, 'population')
# find_duplicate(df2, 'area')
# find_duplicate(df2, 'density')

# col_names = [col for i, col in enumerate(df.columns) if i != 2]
# print(col_names)

# for i in range(len(col_names)):
    # col = col_names[i]
    # print(col)
    # find_duplicate(df2, 'population')

def  str2int(x):
    try:
        return int(x)
    except:
        return 0


df2['population'] = df2['population'].apply(str2int)
df2['density'] = df2['density'].apply(str2int)
df2['density'] = df2['density'].astype('float')
print(df2.info())
print(df2.head(2))
```

```python
def get_city(v:str)->str:
    if(len(v) >= 6):
        return v[:3]
    else:
        return v
    
def get_district(v:str)->str:
    if(len(v) >= 6):
        return v[-3:]
    else:
        return v
    
# print(df2['district'].apply(get_city))
# print(df2['district'].apply(get_district))

df3 = df2
df3['city'] = df3['district'].apply(get_city)
df3['district'] = df3['district'].apply(get_district)
print(df3.head(3))

df3 = df3.reindex(columns = ['city', 'district', 'population', 'area', 'density'])
print(df3.head(3))
```

```python
print(df3.query('population >= 500000'))

pup_max = df3['population'].max()
print(pup_max)

df3.query('population == @pup_max')
```

```python
# calculate sum, mean, std for city's population, area, and density
df4 = df3

# df4.groupby(by='city')[['population', 'area', 'density']].mean()
df4_agg = df4.groupby(by='city')[['population', 'area', 'density']].agg(['sum', 'mean', 'std'])
print(df4_agg.head(3))

print(df4_agg.columns)
print(df4_agg.columns.names)

df4_agg.columns.names = ["data", "statistics"]
print(df4_agg.head(3))
```

```python
# turn into wide dataset
df5 = df4_agg

df5.stack(level=['data', 'statistics'], future_stack=True).unstack(level=['city', 'data'])
```
