```python
import cv2
```

```python
# person
img = cv2.imread('image/face01_1.png', cv2.IMREAD_UNCHANGED)
img = cv2.cvtColor(img, cv2.COLOR_BGR2BGRA)

h = img.shape[0] # height
w = img.shape[1] # width   

# background
bg = cv2.imread('image/white.png', cv2.IMREAD_UNCHANGED)
# bg = cv2.resize(bg,(,h))
bg = cv2.cvtColor(bg, cv2.COLOR_BGR2BGRA)

for x in range(w):
    for y in range(h):
        r = img[y, x, 2]   # Red
        g = img[y, x, 1]   # Green
        b = img[y, x, 0]   # Blue
        if r>90 and r<162 and g>90 and b>105:
            img[y, x] = bg[y, x]   # change to background picture

cv2.imwrite('remove_background.png', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
