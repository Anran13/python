```python
# yolo
# picture preparation
import cv2
```

```python
import random
def rotate_img(name):
    angle = random.randint(0,360)
    img = cv2.imread('yolo/' + name + '.jpg')
    img = cv2.resize(img, (200,200))
    h,w = img.shape[:2]
    center = (w/2, h/2)
    rotate = cv2.getRotationMatrix2D(center, angle=angle, scale=1)
    img_rotate = cv2.warpAffine(img, rotate, (w,h), borderValue=(255,255,255), borderMode=cv2.BORDER_CONSTANT)
    return img_rotate
```
