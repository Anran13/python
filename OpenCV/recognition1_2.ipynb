```python
import cv2
print(cv2.__version__)

import numpy
```

```python
# faces + mosaic
img = cv2.imread('image/mona.png')
grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
face_cascade = cv2.CascadeClassifier("training_model/haarcascade_frontalface_default.xml")
faces = face_cascade.detectMultiScale(grey)
```

```python
for (x, y, w, h) in faces:
    mosaic = img[y:y+h, x:x+w]   # mosaic region
    level = 5                   # mosaic level
    # mh = int(h/level)            # modify the hight according to the mosaic level
    # mw = int(w/level)            # modify the width according to the mosaic level
    mosaic = cv2.resize(mosaic, (mw,mh), interpolation=cv2.INTER_LINEAR) # shrink
    mosaic = cv2.resize(mosaic, (w,h), interpolation=cv2.INTER_NEAREST)  # magnify
    img[y:y+h, x:x+w] = mosaic   # change to mosaic region

cv2.imshow('test', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
