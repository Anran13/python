```python
import cv2
print(cv2.__version__)
```

```python
# turn into grey scale
img = cv2.imread('image/mona.png')
grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
face_cascade = cv2.CascadeClassifier("training_model/haarcascade_frontalface_default.xml")
faces = face_cascade.detectMultiScale(grey)

print(faces)
print(faces.ndim)

# frame the face position with rectangle
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)   

cv2.imshow('test', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# change scaleFactor to 0.7
faces2 = face_cascade.detectMultiScale(grey, scaleFactor=1.1, minNeighbors=5, minSize=(30,30))

# frame the face position with rectangle
for (x, y, w, h) in faces2:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)   

cv2.imshow('test', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
eye_cascade = cv2.CascadeClassifier("training_model/haarcascade_eye.xml")
eyes = eye_cascade.detectMultiScale(grey, minNeighbors=5)
for (x, y, w, h) in eyes:# green rectangle
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2) 

mouth_cascade = cv2.CascadeClassifier("training_model/haarcascade_mcs_mouth.xml") 
mouths = mouth_cascade.detectMultiScale(grey, minNeighbors=5)        
for (x, y, w, h) in mouths:# red rectangle
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 0, 255), 2) 

nose_cascade = cv2.CascadeClassifier("training_model/haarcascade_mcs_nose.xml")
noses = nose_cascade.detectMultiScale(grey, minNeighbors=5)                  
for (x, y, w, h) in noses:# blue rectangle
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)        

cv2.imshow('test', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
